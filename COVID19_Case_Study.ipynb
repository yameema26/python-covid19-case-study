```python
# id_vars=None, value_vars=None, var_name=None, value_name='value'
id_vars = ['Province/State', 'Country/Region', 'Lat', 'Long']
all_cols = confirmed_cases.columns.tolist()
date_cols = [c for c in all_cols if c not in id_vars]
date_cols
```

```python
date_cols = []
for c in all_cols:
    if c not in id_vars:
        date_cols.append(c)
date_cols
```

```python
confirmed_long = confirmed_cases.melt(
    id_vars=id_vars,
    value_vars=date_cols,
    var_name="Date",
    value_name="Confirmed"
)

confirmed_long["Date"] = pd.to_datetime(
    confirmed_long["Date"],
    format="mixed",
    dayfirst=True,
    errors="raise"
)

confirmed_long
```

```python
# Deaths -> Melting -> Unpivot
# id_vars=None, value_vars=None, var_name=None, value_name='value'
id_vars = ['Province/State', 'Country/Region', 'Lat', 'Long']

all_cols = deaths.columns.tolist()
date_cols = [c for c in all_cols if c not in id_vars]

death_long = deaths.melt(
    id_vars=id_vars,
    value_vars=date_cols,
    var_name="Date",
    value_name="Deaths"
)

death_long["Date"] = pd.to_datetime(
    death_long["Date"],
    format="mixed",
    dayfirst=True,
    errors="raise"
)
death_long
```

```python
# recovered_cases -> Melting -> Unpivot
# id_vars=None, value_vars=None, var_name=None, value_name='value'
id_vars = ['Province/State', 'Country/Region', 'Lat', 'Long']
all_cols = recovered_cases.columns.tolist()
date_cols = [c for c in all_cols if c not in id_vars]

recovered_long = recovered_cases.melt(
    id_vars=id_vars,
    value_vars=date_cols,
    var_name="Date",
    value_name = 'Recovered', 
)
recovered_long

recovered_long["Date"] = pd.to_datetime(
    recovered_long["Date"],
    format="mixed",
    dayfirst=True,
    errors="raise"
)
recovered_long
```

```python
# 5.1. Analyze the peak number of daily new cases in Germany, France, and Italy.
# Which country experienced the highest single-day surge, and when did it occur?

target_countries = ['Germany', 'France', 'Italy']
country_date_total = (
    # .loc(rows,cols)
    merged.loc[merged['Country/Region'].isin(target_countries)]
    .groupby(['Country/Region','Date'])['Confirmed'].sum().reset_index()
    .sort_values(['Country/Region','Date'])
)
country_date_total
```

```python
# Cumulative -> daily new confirmed cases within each country -> .diff()

country_date_total['DailyNew'] = (
    country_date_total.groupby('Country/Region')['Confirmed'].diff()
)
country_date_total
```

```python
# Identifying missing values
country_date_total[country_date_total['DailyNew'].isna()]
```

```python
country_date_total['DailyNew'] = (
    country_date_total.groupby('Country/Region')['Confirmed'].diff()
    .fillna(country_date_total['Confirmed'])
)
country_date_total
```

```python
country_date_total['DailyNew'].isna().sum()
```

```python
# Pivot Table
daily_new_matrix = (
    country_date_total.pivot(
        columns = 'Country/Region',
        index = 'Date',
        values = 'DailyNew',
    ).sort_index()
)
daily_new_matrix
```

```python
daily_new_matrix['France'].nlargest(5)
```

```python
daily_new_matrix['Germany'].nlargest(5)
```

```python
daily_new_matrix['Italy'].nlargest(5)
```

```python
daily_new_by_country = {}

# target_countries = ['Germany', 'France', 'Italy']
for country in target_countries:
    if country in daily_new_matrix.columns:
        s = daily_new_matrix[country]
        daily_new_by_country[country] = s

daily_new_by_country
```

```python
# Peaks {Country} : {max_value , date} per date

peaks = {}

for country , series in daily_new_by_country.items():
    max_value = int(series.max())
    max_index = series.idxmax()
    peaks[country] = (max_value , max_index)

peaks
```

```python
# Visualize
plt.figure(figsize = (12,6) , facecolor = '#E8E8E8')
for country , series in daily_new_by_country.items():
    plt.plot(series.index , series.values , label = country)

plt.title('Daily New Confirmed Cases in Germany, France, Italy' , color = 'red')
plt.xlabel('Date')
plt.ylabel('New Confirmed Cases')
plt.ylim(-200000, 200000)
plt.legend()
plt.show()
```

```python
# 5.2 Compare the recovery rates (recoveries/confirmed cases) between Canada and Australia as of December 31,
# 2020. Which country showed better management of the pandemic according to this metric?

countries = ['Canada','Australia']
target_date = pd.Timestamp('2020-12-31')

# .loc[rows,cols]
recovery_rates = (
    merged.loc[(merged['Country/Region'].isin(countries)) & (merged['Date'] == target_date),
    ['Country/Region' , 'Recovered' , 'Confirmed'] # columns
    ].groupby('Country/Region')
    .sum().reset_index()
)
recovery_rates
```
